# Tensors

- Data structure that is very similar to arrays and matrices. Used to encode the inputs and outputs of a model.
- Similar to ndarray from numpy but can use hardware accelaration on GPU or similar.
- Optimized for automatic differentiation.

```python
import numpy as np
import torch

data = [[1, 2], [3, 4]]
numpy_data = np.array(data)
tensor_data = torch.tensor(data)

print("Data:\n", data)
print("Numpy:\n", numpy_data)
print("Tensor:\n", tensor_data)
```

Different ways to create a tensor

```python
shape = (5,1)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

Creating a tensor from numpy or another tensor

```python
import numpy as np
import torch

data = np.array([[1,2,3,4]])
tensor = torch.from_numpy(data)
print("Tensor:\n>", tensor)

tensor_copy = torch.full_like(tensor, fill_value=0.1, dtype=torch.float)
print("Tensor:\n>", tensor_copy)

tensor_copy = torch.rand_like(tensor,  dtype=torch.float)
print("Tensor:\n>", tensor_copy)

tensor_copy = torch.ones_like(tensor)
print("Tensor:\n>", tensor_copy)

tensor_copy = torch.zeros_like(tensor)
print("Tensor:\n>", tensor_copy)
```

Displaying the different attributes of a tensor.

```python
tensor = torch.rand(3,4)

print(f"Data of tensor:\n{tensor.data}")
print(f"Shape of tensor:            {tensor.shape}")
print(f"Datatype of tensor:         {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

Slicing tensors (just like in numpy)

```python
tensor = torch.ones(4, 4)
print(f"First row: {tensor[0,:]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[1:-1,1:-1] = 0
print(tensor)
```

Joining tensors together

```python
tensor = torch.rand((2,2))
t1 = torch.cat([tensor, tensor, tensor], dim=0)
print(t1)
t2 = torch.cat([tensor, tensor, tensor], dim=1)
print(t2)
```

Arithmetical operations can be done different ways.

```python
tensor = torch.rand((2,2))

# This computes the matrix multiplication between two tensors. y1, y2, y3 will have the same value
# ``tensor.T`` returns the transpose of a tensor
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)
y3 = torch.rand_like(y1)
torch.matmul(tensor, tensor.T, out=y3)
print(y1)
print(y2)
print(y3)

print()
# This computes the element-wise product. z1, z2, z3 will have the same value
z1 = tensor * tensor
z2 = tensor.mul(tensor)
z3 = torch.rand_like(tensor)
torch.mul(tensor, tensor, out=z3)
print(z1)
print(z2)
print(z3)
```

Single-element tensors If you have a one-element tensor, for example by aggregating all values of a tensor into one value, you can convert it to a Python numerical value using item():

```python
agg = tensor.sum()
agg_item = agg.item()
print(agg_item, type(agg_item))
```
